```python
df_final = pd.read_csv('dados_credit.csv', sep = ';', encoding='latin-1')
#Visualização
df_final.head(4)
```

```python
df_final['observação'].drop_duplicates().shape[0] #9994 -- Unicidade nos dados
```

## Data Wrangling e Feature engineering

```python
#Padronização - Colocar todas as colunas em minúsculos
df_final.columns = df_final.columns.str.lower()
```

```python
#Renomear coluna
##observação para observacao
df_final.rename(columns = {'observação':'observacao'}, inplace = True)
```

### Criar a variável idade_veiculo

* A partir da variável ano_modelo_veic, eu vou criar a variável chamada **idade_veiculo**. A referência é o ano de 2025.

```python
df_final['idade_veiculo'] = 2025 - df_final['ano_modelo_veic']
```

```python
df_final.info()
```

## Remover as linhas que não tem informação do ALVO (target)

```python
df_final.dropna(subset=['target'], inplace=True)
```

# 1 - Divisão dos dados

```python
x_m6 = df_final.drop('target', axis=1) #conjunto de variáveis explicativas
y_m6 = df_final['target'] #ALVO
```

```python
#A base de treino vai ter 70% dos dados
#A base de teste vai ter 30% dos dados
m6treino_x, m6teste_x, m6treino_y, m6teste_y = train_test_split(x_m6, y_m6, random_state = 1935, test_size = 0.3, stratify = y_m6)
```

## 1.1 - Base de treino

```python
#Base de treino completa
df_treino = pd.concat([m6treino_x, m6treino_y], axis=1)
#Visualização
df_treino.head(4)
```

```python
#Volumetria
df_treino.shape[0] #6.643
```

```python
#Frequência do target
round(100 * df_treino['target'].value_counts(True),2) #2.97
```

### Salvar em CSV

```python
#Salvar em CSV
df_treino.to_csv('df_treino.csv', header=True, index=False, sep=';')
```

## 1.2 - Base de teste

```python
#Base de treino completa
df_teste = pd.concat([m6teste_x, m6teste_y], axis=1)
df_teste.head(4)
```

```python
#Volumetria
df_teste.shape[0] #2848
```

```python
#Frequência do ALVO
round(100 * df_teste['target'].value_counts(True),2) #2.98
```

### Salvar em CSV

```python
#Salvar em CSV
df_teste.to_csv('df_teste.csv', header=True, index=False, sep=';')
```
